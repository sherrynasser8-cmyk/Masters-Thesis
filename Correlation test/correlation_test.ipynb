```python
"""
Correlation test: Fidelity vs Utility vs Privacy
across all 18 experiment-dataset combinations (E1-E6 x 3 datasets).

Source of raw values: Seminar presentation evaluation tables
(Luxury / Privacy / Apparel, slides 44-46) and AHP survey question headers.

Metrics used (one per dimension):
    Fidelity -> Pearson L2   (lower = better fidelity)
    Utility  -> TSTR Accuracy (higher = better utility)
    Privacy  -> DCR           (higher = better privacy)

Run:
    pip install scipy --break-system-packages   # if not already installed
    python3 correlation_test.py
"""

from scipy import stats

# ---------------------------------------------------------------
# Raw data: (Fidelity_L2, Utility_Accuracy, Privacy_DCR) per E1..E6
# ---------------------------------------------------------------
data = {
    "Luxury": [
        (4.29, 0.46, 4.44),   # E1
        (3.34, 0.24, 3.65),   # E2
        (2.37, 0.61, 2.49),   # E3
        (2.71, 0.56, 2.65),   # E4
        (0.59, 0.78, 1.35),   # E5
        (1.21, 0.71, 1.43),   # E6
    ],
    "Privacy": [
        (10.04, 0.72, 4.11),  # E1
        (2.20,  0.79, 3.44),  # E2
        (10.09, 0.72, 4.13),  # E3
        (2.93,  0.74, 3.58),  # E4
        (2.93,  0.75, 3.57),  # E5
        (4.51,  0.71, 3.50),  # E6
    ],
    "Apparel": [
        (3.82, 0.78, 4.51),   # E1
        (1.63, 0.99, 3.53),   # E2
        (2.12, 0.70, 4.53),   # E3
        (3.39, 0.79, 3.98),   # E4
        (1.86, 0.90, 3.22),   # E5
        (1.83, 1.00, 2.05),   # E6
    ],
}

EXP_LABELS = ["E1", "E2", "E3", "E4", "E5", "E6"]

# ---------------------------------------------------------------
# Flatten into three parallel lists (n = 18)
# ---------------------------------------------------------------
fidelity, utility, privacy, labels = [], [], [], []
for dataset, rows in data.items():
    for exp_label, (f, u, p) in zip(EXP_LABELS, rows):
        fidelity.append(f)
        utility.append(u)
        privacy.append(p)
        labels.append(f"{dataset}-{exp_label}")

n = len(fidelity)
print(f"n = {n} experiment-dataset combinations\n")

# ---------------------------------------------------------------
# Print the raw table for manual verification
# ---------------------------------------------------------------
print(f"{'Combo':<15}{'Fidelity(L2)':>14}{'Utility(Acc)':>14}{'Privacy(DCR)':>14}")
for lbl, f, u, p in zip(labels, fidelity, utility, privacy):
    print(f"{lbl:<15}{f:>14.3f}{u:>14.3f}{p:>14.3f}")

# ---------------------------------------------------------------
# Pearson correlation tests
# ---------------------------------------------------------------
r_fid_priv, p_fid_priv = stats.pearsonr(fidelity, privacy)
r_uti_priv, p_uti_priv = stats.pearsonr(utility, privacy)
r_fid_uti,  p_fid_uti  = stats.pearsonr(fidelity, utility)

print("\n--- Pearson correlation results (n = {}) ---".format(n))
print(f"Fidelity(L2) vs Privacy(DCR): r = {r_fid_priv:.3f}, p = {p_fid_priv:.4f}")
print(f"Utility(Acc) vs Privacy(DCR): r = {r_uti_priv:.3f}, p = {p_uti_priv:.4f}")
print(f"Fidelity(L2) vs Utility(Acc): r = {r_fid_uti:.3f}, p = {p_fid_uti:.4f}")

# ---------------------------------------------------------------
# Interpretation helper
# ---------------------------------------------------------------
def interpret(r, p, alpha=0.05):
    sig = "SIGNIFICANT" if p < alpha else "not significant"
    strength = (
        "negligible" if abs(r) < 0.1 else
        "weak" if abs(r) < 0.3 else
        "moderate" if abs(r) < 0.5 else
        "strong" if abs(r) < 0.7 else
        "very strong"
    )
    direction = "positive" if r > 0 else "negative"
    return f"{strength} {direction} correlation, {sig} at alpha={alpha}"

print("\n--- Interpretation ---")
print("Fidelity vs Privacy:", interpret(r_fid_priv, p_fid_priv))
print("Utility  vs Privacy:", interpret(r_uti_priv, p_uti_priv))
print("Fidelity vs Utility :", interpret(r_fid_uti, p_fid_uti))
```
